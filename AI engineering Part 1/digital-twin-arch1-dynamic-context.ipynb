### Step 1: Import Libraraies & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown
from openai import OpenAI
import gradio as gr

load_dotenv()

OpenAI_API_KEY = os.getenv("OPENAI_API_KEY")

if  OpenAI_API_KEY is None:
    raise Exception("OPENAI_API_KEY is missing")
```

### Step 2: Simple UI with AI

```python
def respond_ai(message, history):
  messages = [{"role": "system", "content": "You are a helpful assistant."}] + history + [{"role": "user", "content": message}]
  client = OpenAI(api_key=OpenAI_API_KEY)
  response = client.chat.completions.create(
      model="gpt-4.1-mini",
      messages=messages
    )
  
  reply = response.choices[0].message.content
  return reply


  
```

```python
gr.ChatInterface(fn=respond_ai).launch(inbrowser=True)
```

### Step 5: Dynamic Context Injection

```python
Topic_Context = {
    "2011":"I have started pursuing my bachelor's degree in ECE in India.",
    "cooking":"I love cooking and experimenting with new recipes. I enjoy trying out different cuisines and flavors, and I find it to be a great way to relax and express creativity. Cooking is not just about preparing food for me; it's also about the joy of sharing meals with family and friends, and creating memorable experiences around the dining table.",
    "cars": "I love muscle cars and have a deep appreciation for their design, engineering, and performance. I enjoy learning about different car models, their specifications, and the history behind iconic vehicles. Attending car shows and events is something I look forward to, as it allows me to connect with fellow enthusiasts and stay updated on the latest trends in the automotive world.",
    "fitness": "I believe maintaining physical fitness is important for both health and productivity. I enjoy exercising regularly and learning about nutrition and healthy habits.",
    "travel": "I enjoy traveling and experiencing different cultures, local cuisines, and historical places. Exploring new destinations gives me fresh perspectives and memorable experiences.",
    "music": "I enjoy listening to music while working or relaxing. My taste varies depending on my mood, and I appreciate discovering new artists and genres.",
    "gaming": "I enjoy playing video games as a way to unwind and challenge myself. I appreciate games that offer immersive storytelling, strategic thinking, and engaging gameplay mechanics. Gaming also allows me to connect with friends and the gaming community, sharing experiences and tips."
""
}
```

```python
def respond_ai(message, history):

    #Dynamic Injection Code based on the keywords in the message
    system_message_enhanced = system_message
    for keyword , context in Topic_Context.items():
        if keyword in message.lower():
            system_message_enhanced += "\n\n" + context
       





    #AS Usual code
    messages = [{"role": "system", "content": system_message_enhanced}] + history + [{"role": "user", "content": message}]
    client = OpenAI(api_key=OpenAI_API_KEY)
    response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
    )
  
    reply = response.choices[0].message.content
    return reply
```

```python
gr.ChatInterface(fn=respond_ai).launch(inbrowser=True)
```
