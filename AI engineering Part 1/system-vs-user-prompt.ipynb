### Step 1: Import Libraries and Load the environment Variables


```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
load_dotenv

OPENAI_API_KEY=os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API Key is missing")
else:
    print(OPENAI_API_KEY[:8])

cleint = OpenAI()
```

### Step 2: Call the API


```python
response = cleint.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Hi, My name is Tarun."}]
)


#print(response)      
reply=response.choices[0].message.content


print(reply)
```

### Step 3a: Calling the API with a System Prompt

```python
response = cleint.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},  #system Prmpt
        {"role": "user", "content": "How do I make a tresleches Cake? "}
    ]
)

reply=response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Step 3b: Changing the System Prompt changes the LLM's behaviour 


```python
response = cleint.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are hilarious assitant and talks everything like a joke."},  #system Prmpt
        {"role": "user", "content": "How do I make a tresleches Cake? "}
    ]
)

reply=response.choices[0].message.content
display(Markdown(reply))
```

### Commercial use-case: IKEA Customer Service Agent

```python
question= " I brought a work desk from your store last month. The desk has a broken leg and I would like to request a replacement. Can you please assist me with the process of getting a new desk? Thank you."
```

```python
response = cleint.chat.completions.create(
model="gpt-4o-mini",
messages=[
        {"role": "system", "content": "You are a customer service representative. "
        "you are helpful and friendly but follow strict return policies of the store."
        "Our return policy is: When you return anything, please bring your receipt.​"
        "We issue refunds in the payment form used to buy the product."
        " ​ We’ll refund to an IKEA Refund card to use in-store or online. ​​"
        "More than 365 days since you bought your product? You can still return it as part of our IKEA Buy back service."
        "​If your product is damaged during the delivery process, you can file a claim at IKEA.​  "},  #system Prmpt
        {"role": "user", "content": question}
    ]
)
reply=response.choices[0].message.content
display(Markdown(reply))
```
