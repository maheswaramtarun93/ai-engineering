```python
from openai import OpenAI
from IPython.display import Markdown, display

# Line 1: Initialize the OpenAI client using your API key.
client = OpenAI(api_key="YOUR_API_KEY")

# Line 2: Call the chat completions endpoint to generate a response.
response = client.chat.completions.create(

    # Line 3: Specify which GPT model to use.
    model="gpt-5.5",

    # Line 4: Start the messages list with conversation history.
    messages=[

        # Line 5: System message sets the AI's behavior and context.
        {"role": "system", "content": "You are a helpful AI assistant."},

        # Line 6: User message contains the actual prompt/question.
        {"role": "user", "content": "Explain what artificial intelligence is in simple terms."},
    ]
)

# Line 9: Extract the text content from the first response choice.
answer = response.choices[0].message.content

# Line 10: Display the response as formatted Markdown in the notebook.
display(Markdown(answer))
```
